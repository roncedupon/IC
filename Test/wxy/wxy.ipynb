```python
# 导入设置
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression

# bank_name='工商银行'
bank_name='贵州茅台'
# bank_name='赫美集团'
# bank_name='拓维信息'
bank_name_all=['工商银行','贵州茅台','赫美集团','拓维信息']
# 设置显示中文
# plt.rcParams['font.sans-serif'] = ['KaiTi']
# plt.rcParams['axes.unicode_minus'] = False

for i in range(4):
    # print(i)
    print("=========================",bank_name_all[i],"=================================")
    #读取数据
    bank_name=bank_name_all[i]
    df = pd.read_excel('test.xlsx')
    df.head(10)
    X = df[['投资者悲观情绪指标月均值']]  #这里的自变量必须写成二维形式，一个因变量可能对应多个自变量
    Y = df[bank_name]
    # Y=abs(Y)
    # print(X[:,1],Y)
    # 模型搭建
    model = LinearRegression()
    model.fit(X,Y)
    
    # 模型可视化
    plt.scatter(X,Y)
    plt.plot(X,model.predict(X),color='red')
    plt.xlabel('投资者悲观情绪指标月均值')
    plt.ylabel(bank_name[i])
    plt.show()
    
    #查看系数,截距：线性回归方程构造
    print(bank_name_all[i],"的一元线性回归方程的系数是",model.coef_)
    print(bank_name_all[i],"的一元线性回归方程的截距是",model.intercept_)

    # 计算相关系数

    x = X.to_numpy()[:,0]
    y =Y.to_numpy()
    # print(x,y)
    correlation_matrix = np.corrcoef(x, y)
    correlation_coefficient = correlation_matrix[0, 1]
    print(bank_name_all[i],"的相关系数是",correlation_coefficient)

    #相关系数检验
    r=correlation_coefficient
    print(bank_name_all[i],"的相关系数检验是",r*np.sqrt(12-2)/np.sqrt(1-np.power(r,2)))
   
```

```python
# 导入设置
import matplotlib.pyplot as plt
import pandas as pd
import numpy as np
from sklearn.linear_model import LinearRegression

# bank_name='工商银行'
bank_name='贵州茅台'
# bank_name='赫美集团'
# bank_name='拓维信息'
bank_name_all=['工商银行','贵州茅台','赫美集团','拓维信息']
# 设置显示中文
# plt.rcParams['font.sans-serif'] = ['KaiTi']
# plt.rcParams['axes.unicode_minus'] = False

for i in range(4):
    # print(i)
    print("=========================",bank_name_all[i],"=================================")
    #读取数据
    bank_name=bank_name_all[i]
    df = pd.read_excel('test.xlsx')
    df.head(10)
    X = df[['投资者悲观情绪指标月均值']]  #这里的自变量必须写成二维形式，一个因变量可能对应多个自变量
    Y = df[bank_name]

    x=X.to_numpy()
    y=Y.to_numpy()

    mean=y.sum()/12
    var=np.power(y-mean,2).sum()/12
    Ystd=np.sqrt(var)
    print("y标准差",Ystd)

    mean=x.sum()/12
    var=np.power(x-mean,2).sum()/12
    Xstd=np.sqrt(var)
    print("x标准差",Xstd)
    print("f检验结果",Ystd/Xstd)
```
