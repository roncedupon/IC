### 读取一个二进制文件并读取前8字节


```python
with open("map.bin", 'rb') as file:
    pre_8_bytes = file.read(8)
    print(pre_8_bytes)
    print(type(pre_8_bytes))#字节串类型
    print(pre_8_bytes[0:2])
    print(pre_8_bytes.hex())
    binary_str = ' '.join([format(byte, '08b') for byte in pre_8_bytes])#打印二进制输出
    print(binary_str)
    binary_str = ''.join([format(byte, '08b') for byte in pre_8_bytes])[32:46]#打印二进制输出
    print(binary_str)
    print(int(binary_str,2))
```

### 将一个字节串转换成int
pre_8_bytes[0]返回的是对应字节对应的整数
pre_8_byte[0:1]则是返回一个字节序列，
需要注意的是int.from_bytes()接收的是一个字节序列

```python
pre_8_bytes=b's\x02\x00\x00\x92\t\x00\x00'#注意这里的s也是一个二进制，也就是pre_8_bytes[0]==115-->(s的ascil码)
print(pre_8_bytes[0])
print(pre_8_bytes[0:1])
print(int.from_bytes(pre_8_bytes[0:1],"little"))

print(pre_8_bytes[3])
print(int.from_bytes(pre_8_bytes[3:4],"little"))

```

```python
import numpy as np

# 创建一个示例 NumPy 向量
vector = np.array([123, 45, 6789, 10, 234]).reshape(-1,1)

# 指定输出文件
output_file = 'output.txt'

# 将元素写入txt文件，并右对齐
with open(output_file, 'w') as f:
    for element in vector:
        f.write(f"{str(element).rjust(10)}\n")  # 每个元素占10个字符的宽度，右对齐

print(f"数据已写入到 {output_file}")
```
