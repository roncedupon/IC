```python
import os
import glob

def get_file_or_files(input_path):
    # 检查输入路径是否是一个文件
    if os.path.isfile(input_path):
        # 如果是文件，直接返回文件名
        return [os.path.basename(input_path)]
    # 检查输入路径是否是一个文件夹
    elif os.path.isdir(input_path):
        # 如果是文件夹，返回文件夹下所有文件名（非递归）
        return [os.path.basename(file) for file in glob.glob(os.path.join(input_path, "*"))]
    else:
        raise ValueError(f"输入路径 {input_path} 既不是文件也不是文件夹")

# 示例用法
input_path = "/home/dy/IC/python"  # 输入文件路径或文件夹路径
result = get_file_or_files(input_path)
print(result)
```
