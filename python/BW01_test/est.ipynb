```python
import re
from collections import defaultdict

# 示例字段列表
fields = [
    "AA_bb_cc_0", "AA_bb_cc_1", "AA_bb_cc_2",
    "BB_cc_dd_0", "BB_cc_dd_1", "BB_cc_dd_2",
    "CC_dd_ee_0", "CC_dd_ee_1",
    "A0_bbcc_0", "A0_bbcc_1", "AAA_bbcc", 
    "1_m95v_0", "1_m95v_1", "1_m95v_2", "1_m95v",
    "DAC_50", "DAC_54","DAC_58","DAC_C","ARRAY_18"
]

# 定义正则表达式
pattern = re.compile(r"^(.*?)(?:_(\d+))?$")

# 分类字段
categories = defaultdict(list)

# 对字段进行分类
for field in fields:
    match = pattern.match(field)
    if match:
        prefix = match.group(1)  # 提取前缀部分
        suffix = match.group(2)  # 提取数字后缀（如果有）
        if suffix:
            # 将数字后缀存为整数，方便后续比较
            categories[prefix].append((int(suffix), field))
        else:
            # 没有后缀数字的字段直接归类
            categories[prefix].append((None, field))

# 处理分类结果
result = defaultdict(list)

for prefix, items in categories.items():
    # 按后缀数字排序（None 处理为最小值）
    items.sort(key=lambda x: (x[0] is None, x[0]))

    # 处理连续性
    temp_group = []
    prev_suffix = None

    for suffix, field in items:
        if prev_suffix is not None and suffix is not None:
            # 检查是否连续
            if suffix - prev_suffix > 1:
                # 不连续，将当前分组加入结果并开始新分组
                result[f"{prefix}_{prev_suffix}"].extend(temp_group)
                temp_group = []
        temp_group.append(field)
        prev_suffix = suffix

    # 将最后一组加入结果
    if temp_group:
        result[prefix].extend(temp_group)

# 输出结果
for category, items in result.items():
    print(f"类别: {category}")
    print(f"字段: {items}")
    print()
```

```python
import re

def count_field_types(fields):
    """
    Counts the occurrences of different field types in a list of fields.

    Args:
        fields: A list of field names.

    Returns:
        A dictionary where keys are field types and values are their counts.
    """

    field_types = {}
    for field in fields:
        # Extract the prefix, considering "1_m95v" as a single type
        prefix_match = re.match(r"^[A-Za-z0-9]+(?:_[A-Za-z0-9]+)*|1_m95v", field)
        if prefix_match:
            prefix = prefix_match.group(0)
            field_types[prefix] = field_types.get(prefix, 0) + 1
    return field_types

# Example usage
fields = [
    "AA_bb_cc_0", "AA_bb_cc_1", "AA_bb_cc_2",
    "BB_cc_dd_0", "BB_cc_dd_1", "BB_cc_dd_2",
    "CC_dd_ee_0", "CC_dd_ee_1",
    "A0_bbcc_0", "A0_bbcc_1", "AAA_bbcc",
    "1_m95v_0", "1_m95v_1", "1_m95v_2", "1_m95v"
]

result = count_field_types(fields)
print(result)
```

```python
import re

def get_number_after_xs(string):
    """
    提取字符串中 'xs_' 后的数字。

    Parameters:
    - string: 输入字符串。

    Returns:
    - 提取到的数字（字符串形式），如果未找到，返回 None。
    """
    match = re.search(r'xs_(\d+)', string)
    if match:
        return match.group(1)  # 提取第一个捕获组中的数字
    return None

# 示例
example_string = "some_text_xs_123_xs_456_other_text"
result = get_number_after_xs(example_string)
print("提取的数字:", result)
```

```python
import numpy as np
from openpyxl import Workbook

# 创建示例 NumPy 数组
data = np.random.randint(0, 100, size=(4, 5))

# 创建一个 Workbook 对象
workbook = Workbook()
sheet = workbook.active  # 获取默认工作表

# 将 NumPy 数组写入 Excel 文件
for row_idx, row in enumerate(data, start=1):
    for col_idx, value in enumerate(row, start=1):
        sheet.cell(row=row_idx, column=col_idx, value=value)

# 保存为 Excel 文件
workbook.save("output.xlsx")
print("数组已成功保存到 output.xlsx 文件")
```
