# randome测试
实际项目中发现一个很严重的问题：\
首先numpy.random.seed(12345),然后创建一个class,在这个class中有一个函数funA，这个函数funA只做了一次随机，结束运行\
接下来，继续调用这个class中的该函数两次，之前想的是numpy.random.seed(12345)设置的是全局随机值，然鹅实际发现并不是

```python
import numpy as np
np.random.seed(12345)
class myclass:
    def funA():
        print(np.random.randint(0,255,(10,10)))
myclass_instance=myclass
myclass_instance.funA()
print("--"*10)
myclass_instance.funA()
```

新的问题，现在我们将random.seed挪到函数内部，在函数内部随机两次，并且在外面调用该函数2次，\
可以发现两次函数调用中，只有第一次随机一致，第二次就不一致了

```python
import numpy as np

class myclass:
    def __init__(self):
        np.random.seed(12345)
    def funA():
        # np.random.seed(12345)
        AA=np.random.randint(0,255,(1,10))if True else np.random.randint(0,255,(1,10))

        BB=np.random.randint(0,255,(1,15))if False else np.random.randint(0,255,(1,15))
        print(AA)
        print(BB)
myclass_instance=myclass
myclass_instance.funA()
print("--"*10)
myclass_instance.funA()
```
