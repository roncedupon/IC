```python
#%%
import matplotlib.pyplot as plt

# 创建数据
x = [1, 2, 3, 4, 5]
y = [1, 4, 9, 16, 25]

# 创建图形和坐标轴
fig, ax = plt.subplots()

# 绘制图形
ax.plot(x, y)

# 设置x和y轴的范围，确保包含0
ax.set_xlim(left=0)  # x轴最小值为0
ax.set_ylim(bottom=0)  # y轴最小值为0

# 调整坐标轴spines，使得x轴和y轴从原点开始
ax.spines['left'].set_position('zero')
ax.spines['bottom'].set_position('zero')

# 显示图形
plt.show()
```

```python
import matplotlib.pyplot as plt

# 示例数据 (假设 modules_data 是一个列表，每个元素是 (module, start_time, end_time) 元组)
modules_data = [
    ('Module A', 10, 20),
    ('Module B', 15, 25),
    ('Module C', 5, 15),
    ('Module D', 22, 30),
]

module_to_color = {  # 示例颜色映射
    'Module A': 'skyblue',
    'Module B': 'lightcoral',
    'Module C': 'lightgreen',
    'Module D': 'lightsalmon',
}

fig, ax = plt.subplots(figsize=(8, 6)) # 调整图表大小，可选

for i, (module, start_time, end_time) in enumerate(modules_data):
    duration = end_time - start_time
    color = module_to_color.get(module, 'gray') # 默认颜色为灰色，如果模块没有在字典中

    ax.barh(module, duration, left=start_time, height=0.6, align='center', color=color)

    # 在柱状图左下角添加 start_time 文本
    ax.text(start_time,  # x 坐标：柱状图的左侧起始位置
            i,          # y 坐标：使用 enumerate 的索引 i 作为 y 轴位置
            str(start_time), # 要添加的文本，将 start_time 转换为字符串
            ha='left',    # 水平对齐：左对齐
            va='bottom',  # 垂直对齐：底部对齐
            fontsize=8,   # 字体大小，可以根据需要调整
            color='black') # 文本颜色，可以根据需要调整

ax.set_xlabel('Time')
ax.set_ylabel('Module')
ax.set_title('甘特图示例 - 带起始时间标注')
ax.invert_yaxis()  # 可选：反转 Y 轴，让模块从上到下排列
ax.grid(axis='x', linestyle='--') # 可选：添加 x 轴网格线

plt.show()
```

```python
import matplotlib.pyplot as plt

# 示例数据
modules_data = [
    ('Module A', 10000*10, 10000*10+10),
    ('Module B', 10000*15, 10000*15+10),
    ('Module C', 10000*5, 10000*5+10),
    ('Module D', 10000*22, 10000*22+10),
]

module_to_color = {
    'Module A': 'skyblue',
    'Module B': 'lightcoral',
    'Module C': 'lightgreen',
    'Module D': 'lightsalmon',
}

# 设置最小持续时间，用于图形可见性
min_duration = 100  # 设定最小持续时间为 100（单位是时间戳差异）

fig, ax = plt.subplots(figsize=(8, 6))

for i, (module, start_time, end_time) in enumerate(modules_data):
    duration = end_time - start_time
    color = module_to_color.get(module, 'gray')

    # 如果持续时间太短，调整为最小显示时间
    if duration < min_duration:
        duration = min_duration

    ax.barh(module, duration, left=start_time, height=0.6, align='center', color=color)

    # 在柱状图左下角添加 start_time 文本
    ax.text(start_time, i, str(start_time), ha='left', va='bottom', fontsize=8, color='black')

ax.set_xlabel('Time')
ax.set_ylabel('Module')
ax.set_title('甘特图示例 - 带起始时间标注')

# 设置 x 轴范围，适当设置，使图形更清晰
ax.set_xlim([1e4, 1e7])

ax.invert_yaxis()  # 反转 Y 轴
ax.grid(axis='x', linestyle='--')

plt.show()
```
