# 字典

```python
my_dict = {'a': 1, 'b': 2, 'c': 3}
```

## 获取key

```python
#获取所有key
my_keys=my_dict.keys()
type(my_keys)#这样获得的遍历类型是dict_keys
# 如果需要转换为列表
keys_list = list(my_keys)
keys_list
```

## 获取value

```python
#获取所有的value
my_values=my_dict.values()
#同样地，想要index该元素，还需要转list
values_list=list(my_values)
print(values_list)
```

## 即想要key，也想要value

```python
#想要获取key，和value
my_items=my_dict.items()
item_list=list(my_items)
print(item_list[0])#tuple
print(item_list)#list
print(len(my_items))
```

```python
for key,value in my_dict.items():
    print(key,"   ",value)
```
